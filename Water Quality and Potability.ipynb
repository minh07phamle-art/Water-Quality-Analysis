```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
print("Libraries imported successfully.")
```

```python
data=pd.read_csv('water_potability.csv')
data.head()
```

```python
print(data.isnull().sum())
print(data.shape)
```

```python
data = data.fillna(data.mean())
```

```python
data.describe()
```

```python
sns.histplot(data['ph'], kde=True, color='orange')
plt.title('pH Level in Water samples')
plt.xlabel('pH')
```

```python
plt.figure(figsize=(20, 6))
sns.heatmap(data.corr(),annot=True, cmap='coolwarm')
plt.title('Correlation Between Water Quality Parameters')
```

```python
fig, axes = plt.subplots(2, 3, figsize=(20, 10))

sns.boxenplot(x='Potability', y='ph', data=data, ax=axes[0, 0])
sns.boxenplot(x='Potability', y='Hardness', data=data, ax=axes[0, 1])
sns.boxenplot(x='Potability', y='Chloramines', data=data, ax=axes[1, 0])
sns.boxenplot(x='Potability', y='Sulfate', data=data, ax=axes[1, 1])

plt.suptitle("Paramenters vs Potability")
```
